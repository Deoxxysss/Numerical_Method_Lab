# Theory: Linear Algebra and Systems of Linear Equations

## What is Linear Algebra?
Linear algebra is the branch of mathematics that studies **vectors**, **vector spaces**, **linear transformations**, and **systems of linear equations**. It provides tools to solve problems involving straight lines, planes, and higher‑dimensional analogues.

## Representing Lines and Planes
- In **2D**, a line can be described by a single linear equation:  
  \( ax + by = c \)  
  Alternatively, in vector form:  
  \( \mathbf{r} = \mathbf{a} + \lambda \mathbf{b} \), where \( \mathbf{a} \) is a position vector, \( \mathbf{b} \) is a direction vector, and \( \lambda \in \mathbb{R} \).

- In **3D**, a line is also 1‑dimensional and is best represented parametrically:  
  \( \mathbf{r} = \mathbf{a} + \lambda \mathbf{b} \).  
  A single equation like \( ax + by + cz = d \) defines a **plane**, not a line.

- Systems of linear equations arise when we have multiple such equations. For example:  
  \[
  \begin{cases}
  a_1x + b_1y = c_1 \\
  a_2x + b_2y = c_2
  \end{cases}
  \]
  The solution (if it exists) is the point where the two lines intersect.

## Solving Systems of Linear Equations
The goal is to find values for the variables that satisfy all equations simultaneously. Geometrically, this corresponds to finding the intersection of the corresponding geometric objects. Methods include:
- Gaussian elimination
- Matrix inversion
- Iterative methods (Jacobi, Gauss‑Seidel)

In this notebook, we will implement and compare some of these numerical methods.

# Theory: Root Finding in One Variable

## What is Root Finding?

A **root** of a real‑valued function \( f(x) \) is a value \( x = r \) such that  
\[
f(r) = 0.
\]

Root finding is the process of locating these values numerically when an exact algebraic solution is impossible or impractical. Geometrically, a root corresponds to the point where the graph of \( f(x) \) crosses the \( x \)-axis.

## Why do we need numerical methods?

Many equations cannot be solved exactly. For example, \( e^x - x^2 = 0 \) or \( \cos(x) = x \) have no simple closed‑form solution. Numerical methods provide approximations to any desired accuracy using iterative procedures.

## Common Methods

1. **Bisection Method**  
   - Requires an interval \([a,b]\) where \( f(a) \) and \( f(b) \) have opposite signs (ensuring at least one root by the Intermediate Value Theorem).  
   - Each iteration halves the interval, guaranteeing convergence but relatively slowly (linear convergence).  
   - **Advantage:** robust and always converges if the initial bracket is valid.  
   - **Disadvantage:** slow; requires a valid starting interval.

2. **Newton–Raphson Method**  
   - Uses the tangent line at the current estimate:  
     \[
     x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}.
     \]  
   - Requires the derivative \( f'(x) \), which may be given analytically or approximated numerically.  
   - **Advantage:** very fast (quadratic convergence) when close to the root.  
   - **Disadvantage:** may diverge if the initial guess is poor or if \( f'(x) \) is zero near the root.

3. **Secant Method**  
   - Similar to Newton’s method but approximates the derivative using a finite difference from two previous points.  
   - No need to provide \( f'(x) \), but convergence is slightly slower than Newton’s method (super‑linear).

## Convergence and Stopping Criteria

A numerical method usually stops when one of the following is satisfied:
- The absolute function value is small: \( |f(x_n)| < \text{tol} \).
- The change in \( x \) between iterations is small: \( |x_{n+1} - x_n| < \text{tol} \).
- A maximum number of iterations is reached (to prevent infinite loops).

## Important Concepts

- **Order of convergence** measures how fast the error decreases. Bisection is linear (error reduces by a constant factor each step), while Newton’s method is quadratic (error squares roughly each step).
- **Multiple roots**: If \( f(r) = 0 \) and \( f'(r) = 0 \), the root is multiple, and Newton’s method converges only linearly unless modified.
- **Bracketing vs. open methods**: Bisection is a bracketing method (maintains an interval containing the root). Newton and secant are open methods (only need initial guesses, may not converge).

## Example

Find the root of \( f(x) = x^2 - 2 \) (i.e., \( \sqrt{2} \)) using the bisection method with starting interval \([1,2]\) and tolerance \(10^{-6}\). The root is approximately \(1.41421356237\).

## Common Misconceptions to Avoid

- **“Root” ≠ “solution to an equation”**: A root is specifically the \( x \)-value where the function equals zero. Solving \( f(x) = g(x) \) is equivalent to finding the root of \( f(x) - g(x) \).
- **Bisection method always works**: It requires the function to be continuous and the starting interval to have opposite signs; otherwise, it fails or gives no guarantee.
- **Newton’s method always converges**: It can oscillate, diverge, or hit a point where the derivative is zero, especially with a bad initial guess.
- **Tolerance is the same as accuracy**: The tolerance is the stopping criterion; the actual error may be larger if the function is ill‑conditioned.

# Theory: Numerical Integration

## What is Numerical Integration?

**Numerical integration** (also called quadrature) is the process of approximating the definite integral of a function when an exact analytic antiderivative is impossible or impractical to compute. The integral
\[
\int_a^b f(x)\,dx
\]
represents the area under the curve \( f(x) \) between \( x=a \) and \( x=b \). Numerical methods approximate this area using sums based on a finite number of function evaluations.

## Why do we need numerical methods?

Many functions do not have elementary antiderivatives (e.g., \( e^{-x^2} \), \( \sin(x^2) \)), and even when they do, the antiderivative may be extremely complex. Additionally, in practice, \( f(x) \) may be known only at discrete data points (e.g., from measurements). Numerical integration provides a way to compute approximations that can be made as accurate as desired.

## Common Methods

1. **Rectangle Rules**  
   - **Left rectangle rule**: Approximates the area using rectangles whose height is the function value at the left endpoint of each subinterval.  
   - **Right rectangle rule**: Uses the function value at the right endpoint.  
   - **Midpoint rule** (not covered here) uses the midpoint.  
   - These methods are first‑order accurate (error ~ \( O(h) \)) for left/right, but the midpoint rule is second‑order.

2. **Trapezoidal Rule**  
   - Approximates the function by straight lines between consecutive points, forming trapezoids.  
   - Formula:  
     \[
     \int_a^b f(x)\,dx \approx \frac{h}{2} \left[ f(x_0) + 2\sum_{i=1}^{n-1} f(x_i) + f(x_n) \right],
     \]
     where \( h = (b-a)/n \) and \( x_i = a + i h \).  
   - Error is second‑order: \( O(h^2) \).

3. **Simpson’s Rule**  
   - Approximates the function by quadratic polynomials on pairs of subintervals.  
   - Requires an even number of intervals \( n \).  
   - Formula:  
     \[
     \int_a^b f(x)\,dx \approx \frac{h}{3} \left[ f(x_0) + 4\sum_{i \text{ odd}} f(x_i) + 2\sum_{i \text{ even}} f(x_i) + f(x_n) \right].
     \]  
   - Error is fourth‑order: \( O(h^4) \), making it much more accurate for smooth functions.

## Error and Convergence

- The **step size** \( h = (b-a)/n \) controls accuracy: as \( n \) increases (smaller \( h \)), the approximation generally improves.
- The order of a method indicates how fast the error decreases with \( h \). For example, halving \( h \) in the trapezoidal rule reduces error by roughly a factor of 4; for Simpson’s rule, it reduces by a factor of 16.

## Stopping Criteria / Adaptive Methods

In practice, we may not know the exact error. Common approaches:
- Compare results using different \( n \) (e.g., double \( n \) until the change is small).
- Use adaptive quadrature, which automatically refines the mesh where the function changes rapidly.

## Important Concepts

- **Degree of exactness**: A quadrature rule is exact for polynomials up to a certain degree. For instance, the trapezoidal rule is exact for linear functions, while Simpson’s rule is exact for cubic polynomials.
- **Composite rules**: Splitting the interval into many small subintervals and applying a basic rule on each improves accuracy.
- **Round‑off error**: For very large \( n \), floating‑point round‑off may dominate, so increasing \( n \) beyond a certain point may worsen the result.

## Example

Approximate \( \int_0^1 e^{-x^2}\,dx \) using Simpson’s rule with \( n=10 \). The exact value (to 10 decimal places) is about 0.7468241328. Simpson’s rule gives a result accurate to many decimal places even with a small number of intervals.

## Common Misconceptions to Avoid

- **“More points always mean better accuracy”**: Beyond a certain \( n \), round‑off errors can degrade the result.
- **“All methods are equally accurate”**: Simpson’s rule is much more accurate than trapezoidal or rectangle rules for the same \( n \), especially for smooth functions.
- **“Rectangle rules are useless”**: They are simple to implement and understand, and are a good starting point, though they converge slowly.
- **“Numerical integration gives the exact answer”**: It only provides an approximation; the error depends on the method and the step size.

# How I Solved Linear Systems: Gaussian Elimination

When you have a bunch of equations that all need to be true at the same time, you're dealing with a **system of linear equations**. For example:

\[
\begin{cases}
2x + 3y = 5 \\
4x - y = 1
\end{cases}
\]

Here we're looking for values of \(x\) and \(y\) that make both equations work. Geometrically, that's the point where the two lines cross. In more dimensions, it's like finding the intersection of planes or hyperplanes.

## Putting It Into Matrix Form

Instead of writing out all the equations separately, we can pack everything neatly into matrices:

\[
A\mathbf{x} = \mathbf{b}
\]

where \(A\) is the matrix of coefficients, \(\mathbf{x}\) is the unknown vector, and \(\mathbf{b}\) is the constants vector. For the example above:

\[
A = \begin{pmatrix} 2 & 3 \\ 4 & -1 \end{pmatrix}, \quad
\mathbf{x} = \begin{pmatrix} x \\ y \end{pmatrix}, \quad
\mathbf{b} = \begin{pmatrix} 5 \\ 1 \end{pmatrix}.
\]

## The Main Idea Behind Gaussian Elimination

Gaussian elimination is a systematic way to solve these systems. The idea is to change the system into an equivalent one that's super easy to solve – specifically, one where the matrix is **upper triangular** (all zeros below the diagonal). Then we can just solve from the bottom up, a process called **back substitution**.

We can do this because certain operations don't change the solution: swapping rows, multiplying a row by a non‑zero constant, and adding/subtracting a multiple of one row to another. It's like simplifying an equation but with entire rows.

## Step by Step: What My Code Does

I wrote a function called `gaussian_elimination(A, b)` that does exactly this. Here's what happens inside:

### 1. Forward Elimination – Turn the matrix into upper triangular form

For each column from left to right:

- **Partial Pivoting**: First, I find the row below (or at) the current position with the largest absolute value in that column. Then I swap that row with the current row. Why? Because if the number on the diagonal (the pivot) is very small or zero, dividing by it can cause huge errors or even crash. Picking the largest value keeps things stable.
- If after pivoting the pivot is still effectively zero (smaller than \(10^{-12}\)), then the matrix is **singular** – meaning the system doesn't have a unique solution. I raise an error in that case.
- For each row below the pivot, I calculate a multiplier: \(m = \frac{A[j,i]}{A[i,i]}\). Then I subtract \(m\) times the pivot row from that row. This zeroes out the entry below the pivot. I do the same operation on the \(\mathbf{b}\) vector so everything stays consistent.

### 2. Back Substitution – Solve from the bottom up

Now the matrix is upper triangular, so I can solve for the variables starting from the last row and working upward.

For row \(i\) from bottom to top, the value of \(x_i\) is:

\[
x_i = \frac{b_i - \sum_{k=i+1}^{n-1} A[i,k] \cdot x_k}{A[i,i]}
\]

That just means: take the right‑hand side, subtract the already‑known \(x\)'s multiplied by their coefficients, and divide by the coefficient of the unknown I'm solving for.

## Why Partial Pivoting Matters

Imagine you have a system where the first equation is something like \(0x + 1y = 2\). If you don't pivot, you'll try to divide by zero and the algorithm breaks. Pivoting swaps rows so that the largest value in the column becomes the pivot. In my code, I always pivot before eliminating. That way even if the matrix has a zero on the diagonal, it can still be solved as long as it's not singular.

## Complexity: Is It Fast?

Gaussian elimination does about \(O(n^3)\) operations for the elimination part and \(O(n^2)\) for back substitution. For small to medium‑sized systems (say, up to a few hundred equations), it's very fast. For enormous systems, other methods (like iterative ones) might be better, but for my project this is more than enough.

## Example: Solving a Small System

Let's solve:

\[
\begin{aligned}
2x + y &= 5 \\
x + 3y &= 6
\end{aligned}
\]

In matrix form:

\[
A = \begin{pmatrix} 2 & 1 \\ 1 & 3 \end{pmatrix}, \quad
\mathbf{b} = \begin{pmatrix} 5 \\ 6 \end{pmatrix}.
\]

No pivoting needed because 2 is already the largest in the first column. I eliminate the \(x\) in the second row: multiply the first row by \(0.5\) and subtract from the second row. That gives a new second row \(0, 2.5\) and new \(b_2 = 3.5\). Now back substitute: \(2.5y = 3.5\) → \(y = 1.4\). Then from the first equation: \(2x + 1.4 = 5\) → \(x = 1.8\).

So the solution is \(x = 1.8\), \(y = 1.4\).

## A Few Things People Often Get Wrong

- **Gaussian elimination gives exact answers?** No, it's a numerical method and floating‑point rounding can introduce tiny errors. For most practical problems, the error is negligible if pivoting is used.
- **Pivoting is optional?** Definitely not. Without it, you can get wildly wrong results or division by zero.
- **A zero pivot means no solution?** If after pivoting the pivot is zero, the matrix is singular, meaning the system has either no solution or infinitely many solutions. My code raises an error in that case because a unique solution doesn't exist.
- **The matrix must be square?** For my implementation, yes – I solve \(n\) equations for \(n\) unknowns. Rectangular systems need different approaches, but that's beyond this project.

## Wrapping Up

Gaussian elimination with partial pivoting is one of the most fundamental algorithms in numerical linear algebra. It's reliable, straightforward to implement, and works great for small to medium systems. I tested my implementation against NumPy's `linalg.solve` on several random systems, and it matched within floating‑point tolerance. The full code is in `src/linear_algebra.py`.

# Root Finding: Bisection and Newton‑Raphson

When you have an equation like \(x^2 - 2 = 0\), finding the value of \(x\) that makes it true is called **root finding**. A root is a number \(r\) such that \(f(r) = 0\). These two methods – bisection and Newton‑Raphson – are classic ways to approximate roots numerically. I implemented both in `src/root_finding.py`.

## Bisection Method: Slow but Reliable

The bisection method is like a game of "higher or lower". You start with two points \(a\) and \(b\) where the function has opposite signs (one positive, one negative). By the Intermediate Value Theorem, if the function is continuous, there must be at least one root between them.

**How it works step by step:**

1. Compute the midpoint \(c = \frac{a+b}{2}\).
2. Evaluate \(f(c)\). If \(f(c)\) is close enough to zero (or the interval is tiny enough), stop and return \(c\).
3. Otherwise, check which half still has opposite signs: if \(f(a) \cdot f(c) < 0\), the root lies in \([a,c]\); otherwise, it's in \([c,b]\). Update \(a\) or \(b\) accordingly.
4. Repeat until you're happy.

Because each step cuts the interval in half, the method is guaranteed to converge if the initial bracket is valid, but it's relatively slow (linear convergence). It's very robust though – it never jumps around, and it's easy to understand.

**My implementation** in `bisection_method(f, a, b, tol, max_iter)` does exactly that. It first checks that \(f(a)\) and \(f(b)\) have opposite signs; if not, it raises a `ValueError`. Then it loops up to `max_iter` times, halving the interval each time. The stopping condition is when the function value is small or the interval width is smaller than `tol`. If it runs out of iterations, it returns the midpoint of the last interval.

## Newton‑Raphson Method: Fast but Needs a Derivative

Newton‑Raphson is a completely different approach. Instead of narrowing an interval, it uses the **tangent line** at the current guess to zoom in on the root.

**The idea:**

- Start with an initial guess \(x_0\).
- At \(x_n\), compute the tangent line to \(f(x)\). The tangent line crosses the x‑axis at:
  \[
  x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}.
  \]
- This new point \(x_{n+1}\) becomes the next guess.
- Repeat until the change is very small or the function value is near zero.

Geometrically, you're sliding down the tangent toward the root. If the function is smooth and you start close enough, the convergence is **quadratic** – the number of correct digits roughly doubles each step. That's much faster than bisection.

**But there's a catch:** you need the derivative \(f'(x)\). For some functions, that's easy to write by hand; for others, it might be annoying or impossible to get an analytic expression. That's why I made a separate `derivative` function.

# Numerical Integration: Approximating Areas

When you want to find the area under a curve \(f(x)\) between two points \(a\) and \(b\), you're computing the definite integral:

\[
\int_a^b f(x)\,dx.
\]

In calculus class, you often find an antiderivative and evaluate it at the endpoints. But many functions don't have nice antiderivatives (like \(e^{-x^2}\) or \(\sin(x^2)\)), and sometimes you only know the function at a few measured points, not as a formula. That's where **numerical integration** (also called quadrature) comes in. It gives us a way to approximate that area using just a bunch of function values.

I implemented four classic methods in `src/integration.py`: left rectangle, right rectangle, trapezoidal, and Simpson's rule. All of them share the same basic idea: chop the interval into many small pieces, approximate the area of each piece with something simple (a rectangle, a trapezoid, or a parabola), and add them up.

## The Simplest Idea: Rectangle Rules

Imagine splitting the interval \([a,b]\) into \(n\) equal subintervals of width \(h = \frac{b-a}{n}\). The points are \(x_0 = a, x_1, x_2, \ldots, x_n = b\). For each subinterval \([x_i, x_{i+1}]\), we can approximate the area by a rectangle whose height is the function value at one endpoint.

- **Left rectangle rule**: height = \(f(x_i)\)
- **Right rectangle rule**: height = \(f(x_{i+1})\)

So the area of each rectangle is \(f(x_i) \cdot h\) (or \(f(x_{i+1}) \cdot h\)), and the total approximation is the sum of all those little rectangles.

Why is this even reasonable? Because if the function is nearly constant over each small interval, the rectangle gives a decent estimate. As \(n\) gets larger (and \(h\) smaller), the rectangles become thinner and the approximation gets better.

But there's a catch: these rules are not super accurate. If the function is increasing, the left rectangles underestimate the true area and the right rectangles overestimate it. For decreasing functions, it's the opposite. The error decreases only linearly with \(h\) (first‑order accuracy). That means to get one extra digit of accuracy, you might need ten times as many intervals.

## Trapezoidal Rule: A Slight Improvement

Instead of rectangles, we can use trapezoids. A trapezoid's top edge is a straight line connecting \(f(x_i)\) and \(f(x_{i+1})\). This is a much better approximation for a smooth function because a straight line can follow the curve more closely than a flat rectangle.

The area of one trapezoid is:

\[
\frac{h}{2} \left( f(x_i) + f(x_{i+1}) \right).
\]

When you add them all up, the formula simplifies to:

\[
\int_a^b f(x)\,dx \approx \frac{h}{2} \left[ f(x_0) + 2\sum_{i=1}^{n-1} f(x_i) + f(x_n) \right].
\]

The endpoints are counted once, and all the interior points are counted twice because they are shared between two trapezoids.

The trapezoidal rule is second‑order accurate, meaning the error drops like \(h^2\). Halving \(h\) reduces the error by about a factor of 4. For many practical purposes, that's already quite good.

## Simpson's Rule: Parabolas for Even Better Accuracy

Simpson's rule is a step up again. Instead of approximating the function with straight lines, it uses **quadratic polynomials** (parabolas) on pairs of subintervals. A parabola can match the curvature of \(f(x)\) much better than a straight line or a rectangle.

For Simpson's rule, the number of intervals \(n\) must be even. The formula is:

\[
\int_a^b f(x)\,dx \approx \frac{h}{3} \left[ f(x_0) + 4\sum_{i \text{ odd}} f(x_i) + 2\sum_{i \text{ even}} f(x_i) + f(x_n) \right].
\]

The pattern of coefficients is 1, 4, 2, 4, 2, …, 4, 1. This comes from fitting a quadratic through every three consecutive points and integrating exactly.

Simpson's rule is **fourth‑order accurate** – the error drops like \(h^4\). That means if you double \(n\), the error decreases by a factor of 16. For smooth functions, Simpson's rule can give very high accuracy with a modest number of intervals. It's often the go‑to method for simple integration tasks.

## Why These Methods Work: A Bit of Theory

All these rules are examples of **Newton–Cotes formulas**, which are based on interpolating the function with a polynomial and then integrating that polynomial exactly.

- Rectangle rules use a polynomial of degree 0 (constant).
- Trapezoidal rule uses degree 1 (linear).
- Simpson's rule uses degree 2 (quadratic).

The higher the polynomial degree, the more accurately the method can capture the shape of \(f(x)\), and therefore the faster the error decreases as you increase \(n\).

Another important concept is **degree of exactness**. A quadrature rule is exact for polynomials up to a certain degree. For example:
- Trapezoidal rule is exact for linear functions.
- Simpson's rule is exact for cubic polynomials (yes, even though it uses parabolas, it's exact for cubics due to symmetry).

That's why Simpson's rule is so much better than trapezoidal for the same \(n\).

## How I Implemented It

In my code (`src/integration.py`), I wrote four functions:

- `rectanglel(f, a, b, n)` – left rectangles
- `rectangler(f, a, b, n)` – right rectangles
- `trapzoidal(f, a, b, n)` – trapezoidal rule
- `simpson(f, a, b, n)` – Simpson's rule

Each function takes the function \(f\), the limits \(a\) and \(b\), and the number of subintervals \(n\) (default 10000). They all use NumPy to make the calculations fast and clean.

For example, in `trapzoidal`, I first create an array of \(x\) values using `np.linspace(a, b, n+1)`. Then I compute \(f(x)\) for all points at once. The trapezoidal sum is computed using vectorized operations: I add the first and last function values, double the interior values, and multiply by \(h/2\). This is much faster than a Python loop.

In `simpson`, I check if \(n\) is odd and bump it up by one to make it even. Then I compute the weighted sum using slicing tricks: `f(x[1:-1:2])` grabs the odd‑indexed points, and `f(x[2:-1:2])` grabs the even‑indexed interior points. The coefficients 4 and 2 are applied accordingly.

One thing I noticed: Simpson's rule is so accurate that with just \(n=100\) on \(e^x\) over \([0,1]\), the error is already at the level of floating‑point rounding. That shows how powerful higher‑order methods can be.

## Choosing the Right Method

- If the function is known to be piecewise constant or you only have rectangle approximations, use rectangle rules.
- For most smooth functions, the trapezoidal rule is a good balance of simplicity and accuracy.
- Simpson's rule is the best choice when you need high accuracy without too many function evaluations.

But remember: more intervals is not always better beyond a certain point because of round‑off error. For very large \(n\), the sum of many tiny numbers can lose precision. In practice, you rarely need more than a few thousand intervals for typical problems.

Now we'll play with the root-finding functions we implemented: `bisection_method` and `Newton_raphson_method`. We'll also see how the numerical derivative `derivative` helps Newton's method work without an explicit derivative formula.

We'll test them on a few standard functions, compare their accuracy, and see how fast they converge.

````python
import numpy as np
import time
from numerical_method_lab.root_finding import bisection_method, Newton_raphson_method, derivative

def check_root(f, root, true_root, tol=1e-6):
    error = abs(root - true_root)
    print(f"  Approx root: {root:.10f}")
    print(f"  True root:   {true_root:.10f}")
    print(f"  Error:       {error:.2e}")
    if error < tol:
        print("  ✓ Passed")
    else:
        print("  ✗ Failed")
    print()

# --- Test 1: Bisection on f(x)=x^2-2 ---
print("Bisection method on f(x) = x^2 - 2, interval [1,2]:")
f = lambda x: x**2 - 2
true_root = np.sqrt(2)
root_b = bisection_method(f, 1, 2, tol=1e-8)
check_root(f, root_b, true_root, tol=1e-8)

# --- Test 2: Newton with numerical derivative ---
print("Newton–Raphson on f(x) = x^2 - 2, initial guess 1.5 (using numerical derivative):")
root_n = Newton_raphson_method(f, 1.5, tol=1e-8)
check_root(f, root_n, true_root, tol=1e-8)

# --- Test 3: Newton with analytic derivative ---
print("Newton–Raphson with analytic derivative:")
df = lambda f, x: 2*x   # signature df(f, x), ignore first argument
root_n2 = Newton_raphson_method(f, 1.5, df=df, tol=1e-8)
check_root(f, root_n2, true_root, tol=1e-8)

# --- Timing comparison ---
print("Timing comparison (tol=1e-12):")
start = time.time()
root_b = bisection_method(f, 1, 2, tol=1e-12)
time_b = time.time() - start

start = time.time()
root_n = Newton_raphson_method(f, 1.5, tol=1e-12)
time_n = time.time() - start

print(f"  Bisection time: {time_b:.6f} seconds")
print(f"  Newton time:    {time_n:.6f} seconds\n")

# --- Test 4: Another function f(x)=x^3-x-2 ---
print("Bisection on f(x) = x^3 - x - 2, interval [1,2]:")
f2 = lambda x: x**3 - x - 2
true_root2 = 1.5213797068
root_b2 = bisection_method(f2, 1, 2, tol=1e-8)
check_root(f2, root_b2, true_root2, tol=1e-8)

print("Newton–Raphson (numerical derivative) on f(x) = x^3 - x - 2:")
root_n2 = Newton_raphson_method(f2, 1.5, tol=1e-8)
check_root(f2, root_n2, true_root2, tol=1e-8)
````

# Let's Test Our Integration Functions

Now we'll verify that our integration routines work correctly. We'll test them on functions with known integrals, check that the errors behave as expected, and confirm that the methods are implemented properly.

We'll test:
- **Trapezoidal rule** on a linear function (should be exact).
- **Simpson's rule** on quadratic and cubic functions (should be exact for low n).
- **Rectangle rules** – left and right – to see their bias for increasing functions.
- **Convergence** – error should decrease as n increases.
- **Edge cases** – small n like n=1.

Let's run the tests.

````python
import numpy as np
from numerical_method_lab.integration import trapzoidal, rectanglel, rectangler, simpson

# Define test functions
f_linear = lambda x: 2*x + 3
f_quadratic = lambda x: x**2 - 4*x + 5
f_cubic = lambda x: x**3
f_exp = np.exp
f_sin = np.sin

# Exact integrals
exact_linear = 10.0                  # ∫₀² (2x+3) dx = 10
exact_quadratic = 14/3               # ∫₀² (x²-4x+5) dx = 14/3
exact_cubic = 4.0                    # ∫₀² x³ dx = 4
exact_exp_01 = np.e - 1              # ∫₀¹ eˣ dx = e-1
exact_sin_01 = 1 - np.cos(1)         # ∫₀¹ sin x dx = 1-cos(1)

# Helper to report pass/fail
def check(test_name, condition):
    if condition:
        print(f"✓ {test_name}")
    else:
        print(f"✗ {test_name}")

print("Running integration tests...\n")

# Test 1: Trapezoidal exact for linear
check("Trapezoidal exact for linear", 
      np.isclose(trapzoidal(f_linear, 0, 2, 10), exact_linear, rtol=1e-10))

# Test 2: Simpson exact for quadratic (n=2)
check("Simpson exact for quadratic (n=2)", 
      np.isclose(simpson(f_quadratic, 0, 2, 2), exact_quadratic, rtol=1e-10))

# Test 3: Simpson exact for cubic (n=2)
check("Simpson exact for cubic (n=2)", 
      np.isclose(simpson(f_cubic, 0, 2, 2), 4.0, rtol=1e-10))

# Test 4: Left rectangle underestimates increasing function
check("Left rectangle underestimates increasing function", 
      rectanglel(f_exp, 0, 1, 100) < exact_exp_01)

# Test 5: Right rectangle overestimates increasing function
check("Right rectangle overestimates increasing function", 
      rectangler(f_exp, 0, 1, 100) > exact_exp_01)

# Test 6: Trapezoidal error decreases with n
err_trap_100 = abs(trapzoidal(f_sin, 0, 1, 100) - exact_sin_01)
err_trap_1000 = abs(trapzoidal(f_sin, 0, 1, 1000) - exact_sin_01)
check("Trapezoidal error decreases with n", err_trap_1000 < err_trap_100)

# Test 7: Simpson error decreases with n
err_simp_100 = abs(simpson(f_sin, 0, 1, 100) - exact_sin_01)
err_simp_200 = abs(simpson(f_sin, 0, 1, 200) - exact_sin_01)
check("Simpson error decreases with n", err_simp_200 < err_simp_100)

# Test 8: Simpson auto-adjusts odd n
check("Simpson auto-adjusts odd n", 
      np.isclose(simpson(f_sin, 0, 1, 101), simpson(f_sin, 0, 1, 102), rtol=1e-10))

# Test 9: n=1 works for all methods
check("n=1 works for trapezoidal", np.isfinite(trapzoidal(f_linear, 0, 1, 1)))
check("n=1 works for left rectangle", np.isfinite(rectanglel(f_linear, 0, 1, 1)))
check("n=1 works for right rectangle", np.isfinite(rectangler(f_linear, 0, 1, 1)))
check("n=1 works for Simpson (auto-even)", np.isfinite(simpson(f_linear, 0, 1, 1)))

# Test 10: All methods approximate e^x with large n
check("All methods approximate e^x with large n", 
      all([
          np.isclose(trapzoidal(f_exp, 0, 1, 10000), exact_exp_01, rtol=1e-6),
          np.isclose(rectanglel(f_exp, 0, 1, 10000), exact_exp_01, rtol=1e-4),
          np.isclose(rectangler(f_exp, 0, 1, 10000), exact_exp_01, rtol=1e-4),
          np.isclose(simpson(f_exp, 0, 1, 10000), exact_exp_01, rtol=1e-10)
      ]))

# Test 11: Left and right rectangles converge to same value with large n
check("Left and right rectangles converge with large n", 
      np.isclose(rectanglel(f_sin, 0, 1, 100000), rectangler(f_sin, 0, 1, 100000), rtol=1e-5))

# Test 12: Simpson more accurate than trapezoidal for same n
err_trap = abs(trapzoidal(f_sin, 0, 1, 100) - exact_sin_01)
err_simp = abs(simpson(f_sin, 0, 1, 100) - exact_sin_01)
check("Simpson more accurate than trapezoidal for same n", err_simp < err_trap)

print("\nAll tests completed.")
````
